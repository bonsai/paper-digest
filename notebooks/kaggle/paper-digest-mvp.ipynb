## 2. Corpus manifest and hashes\n\nThis step records file identity only. It does not claim that the PDFs have been parsed or that the corpus is representative.\n

```python
def sha256_file(path: Path) -> str:\n    digest = hashlib.sha256()\n    with path.open('rb') as stream:\n        for block in iter(lambda: stream.read(1024 * 1024), b''):\n            digest.update(block)\n    return digest.hexdigest()\n\nmanifest = pd.DataFrame([\n    {'file_name': p.name, 'path': str(p), 'bytes': p.stat().st_size, 'sha256': sha256_file(p)}\n    for p in pdf_paths\n])\nmanifest.to_csv(OUTPUT_DIR / 'pdf_manifest.csv', index=False)\nmanifest\n
```
